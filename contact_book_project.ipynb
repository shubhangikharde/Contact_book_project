```python
#26 August 2026
#Load Modules
import os
import json
```

```python
data_file='contact_info.txt'

#Load Contacts
def load_contacts():
    if not os.path.exists(data_file):
        return []
    with open(data_file,'r') as f:
        return json.load(f)

#Save contacts
def save_contacts(contacts):
    with open(data_file,'w') as f:
        json.dump(contacts,f,indent=4)
        
#Create Contact
def create_contact():
    name=input("Enter Name:")
    phone=input("Enter Contact number:")
    email=input("Enter Mail ID:")
    contact={"Names:":name,"Contact_Num:":phone,"Mail_ID:":email}
    contacts=load_contacts()
    contacts.append(contact)
    save_contacts(contacts)
    print("Contact save success!!")

#Read contacts
def read_contact():
    contacts=load_contacts()
    if not contacts:
        print("No contact found")
        return
    for ind,contact in enumerate(contacts,start=1):
        print(f"Contact Number:{ind}")
        print(f"Contact Name:{contact['Names:']}")
        print(f"Contact Number:{contact['Contact_Num:']}")
        print(f"Contact Email ID:{contact['Mail_ID:']}")
# Update Contact
def update_contact():
    contacts = load_contacts()

    if not contacts:
        print("No contact found")
        return

    read_contact()

    num = int(input("Enter Contact Number to update: "))

    if num < 1 or num > len(contacts):
        print("Invalid Contact Number")
        return

    contact = contacts[num - 1]

    name = input("Enter New Name: ")
    phone = input("Enter New Contact Number: ")
    email = input("Enter New Mail ID: ")

    contact["Names:"] = name
    contact["Contact_Num:"] = phone
    contact["Mail_ID:"] = email

    save_contacts(contacts)

    print("Contact updated successfully!!")


# Delete Contact
def delete_contact():
    contacts = load_contacts()

    if not contacts:
        print("No contact found")
        return

    read_contact()

    num = int(input("Enter Contact Number to delete: "))

    if num < 1 or num > len(contacts):
        print("Invalid Contact Number")
        return

    contacts.pop(num - 1)

    save_contacts(contacts)

    print("Contact deleted successfully!!")

#Design a menu
def menu():
    while True:
        print("\n Contact Details are as follows:")
        print("\n 1:Create Contact")
        print("\n 2:View Contact")
        print("\n 3:Update Contact")
        print("\n 4:Delete Contact")
        print("\n 5:Exit")

        choice=input("Enter a your choice:")
        if choice=='1':
            create_contact()
        elif choice=='2':
            read_contact()
        elif choice=='3':
            update_contact()
        elif choice=='4':
            delete_contact()
            
        elif choice=='5':
            print("Thank You!!")
            break
        else:
            print("Invalid Choice")
            
```

```python
#Call the method
menu()
```
